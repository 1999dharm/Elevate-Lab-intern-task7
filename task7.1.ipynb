```python
import sqlite3
import pandas as pd
```

```python
# Load CSV
df = pd.read_csv("amazon.csv")
```

```python
# connect to sqlite
conn = sqlite3.connect('sales_data.db')
```

```python
# Save to database
df.to_sql("sales",conn,if_exists ="replace",index=False)
print("Csv data imported successfully")
```

```python
# Clean 'discounted_price' column (remove ₹ and commas)
df["discounted_price"] = df["discounted_price"].str.replace("₹","").str.replace(',','').astype(float)
```

```python
# Clean 'rating_count' column (remove commas and convert to int
df['rating_count'] = df['rating_count'].str.replace(',', '', regex=False)
```

```python
# Step 2: Convert to numeric (int), handling missing values if any
df['rating_count'] = pd.to_numeric(df['rating_count'], errors='coerce').fillna(0).astype(int)
```

```python
# Optional: Check the cleaned column
print(df['rating_count'].head())
```

```python
# Drop missing rows
df.dropna(subset=['product_name', 'discounted_price', 'rating_count'], inplace=True)
```

```python
# Rename columns for SQL compatibility
df.rename(columns={
    'product_name': 'product',
    'discounted_price': 'price',
    'rating_count': 'quantity'
}, inplace=True)
```

```python
# Step 3: Calculate estimated revenue
df['revenue'] = df['quantity'] * df['price']
```

```python
# Step 4: Save cleaned data to SQLite
conn = sqlite3.connect("sales_data.db")
df[['product', 'quantity', 'price', 'revenue']].to_sql("sales", conn, if_exists='replace', index=False)
```

```python
# Step 5: SQL query to summarize sales
query = """
SELECT 
    product, 
    SUM(quantity) AS total_qty, 
    ROUND(SUM(revenue), 2) AS revenue
FROM sales
GROUP BY product
ORDER BY revenue DESC
LIMIT 10
"""

summary_df = pd.read_sql_query(query, conn)
conn.close()
```

```python
# Step 6: Display the summary using print
print("=== Basic Sales Summary (Top 10 Products by Revenue) ===")
print(summary_df)
```

```python
import matplotlib.pyplot as plt
```

```python
# Step 7: Plot bar chart
plt.figure(figsize=(8, 4))
plt.bar(summary_df['product'], summary_df['revenue'], color='skyblue')
plt.title("Top 10 Products by Revenue")
plt.xlabel("Product")
plt.ylabel("Revenue (₹)")
plt.xticks(rotation=30, ha='right')
plt.tight_layout()
plt.savefig("sales_chart.png")
plt.show()
```

```python
# Step 3: Scatter plot for quantity sold vs revenue
plt.figure(figsize=(10, 6))
plt.scatter(df['quantity'], df['revenue'], alpha=0.6, color='orange')
plt.title("Quantity Sold vs Revenue")
plt.xlabel("Quantity Sold")
plt.ylabel("Revenue (₹)")
plt.tight_layout()
plt.savefig("quantity_vs_revenue.png")
plt.show()
```

```python
# Step 4: Revenue distribution histogram
plt.figure(figsize=(10, 6))
plt.hist(df['revenue'], bins=30, color='blue', alpha=0.7)
plt.title("Revenue Distribution of Products")
plt.xlabel("Revenue (₹)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig("revenue_distribution.png")
plt.show()
```

```python
# Step 5: Pie chart for top products' revenue share
top_products = summary_df.head(5)
plt.figure(figsize=(12, 6))
plt.pie(top_products['revenue'], labels=top_products['product'], autopct='%1.1f%%', startangle=90, colors=plt.cm.Paired.colors)
plt.title("Top 5 Products by Revenue Share")
plt.axis('equal')  # Equal aspect ratio ensures the pie chart is circular.
plt.savefig("revenue_share_pie_chart.png")
plt.show()
```
